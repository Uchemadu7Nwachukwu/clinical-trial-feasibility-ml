```python
import pandas as pd
import numpy as np

# Make the results reproducible
np.random.seed(42)

# Number of clinical trial sites
n_sites = 100

# Create the site data
data = {
    "site_id": range(1, n_sites + 1),
    "site_experience": np.random.randint(0, 20, n_sites),
    "patient_pool": np.random.randint(1000, 10000, n_sites),
    "startup_speed": np.random.randint(30, 90, n_sites)
}

# Create DataFrame
df = pd.DataFrame(data)

# Create our prediction target
df["enrollment_days"] = (
    (df["startup_speed"] * 2)
    + (10000 / df["patient_pool"]) * 150
    - (df["site_experience"] * 2)
    + np.random.normal(0, 10, n_sites)
)

# Make enrollment days whole numbers
df["enrollment_days"] = df["enrollment_days"].round().astype(int)

# Make sure enrollment days are at least 30
df["enrollment_days"] = df["enrollment_days"].clip(lower=30)

# Save the dataset
df.to_csv("../data/clinical_trial_sites.csv", index=False)

# Display the first 10 rows
print(df.head(10))

print("\nDataset shape:", df.shape)
print("Dataset saved to data/clinical_trial_sites.csv")
```
